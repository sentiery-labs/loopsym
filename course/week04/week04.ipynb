# 4주차 — 다이어그램 = 그래프, 실행 순서 = 위상 정렬

**이번 주에 완성되는 엔진 조각**
```
블록(완성)  →  [Diagram: connect 검증 + compile 위상정렬]  →  솔버  →  폐루프
               ▲ 여기
```

## 이론 요약
- 다이어그램 = 그래프: 블록이 노드, 와이어가 간선
- 와이어 자료구조: `{(dst블록, dst포트): (src블록, src포트)}`
- 불변식: **입력 포트당 드라이버는 하나** — 위반 시도는 connect 에서 즉시 에러
- 실행 순서: "입력을 주는 블록이 먼저" → 의존 그래프 → Kahn 위상 정렬

````python
import importlib, myblocks, mydiagram
importlib.reload(myblocks); importlib.reload(mydiagram)
bd = mydiagram.Diagram()
c = bd.add(myblocks.Constant(1.0, name='c'))
a = bd.add(myblocks.Gain(2.0, name='a')); b = bd.add(myblocks.Gain(3.0, name='b'))
s = bd.add(myblocks.Sum('++', name='s')); sc = bd.add(myblocks.Scope(name='sc'))
bd.connect(c, a); bd.connect(c, b)          # 다이아몬드!
bd.connect(a, s[0]); bd.connect(b, s[1]); bd.connect(s, sc)
print('실행 순서:', [blk.name for blk in bd.compile()])
````
