**이번 주에 완성되는 엔진 조각**
```
신호 함수  →  [블록 = 파라미터 + output()]  →  다이어그램  →  솔버  →  폐루프
               ▲ 여기 (Gain, Saturation, Offset)
```
목표: "블록 = 상태 + 동작을 가진 객체". 클래스/인스턴스/상속.

## 이론 요약
- 왜 함수가 아니라 클래스인가? **파라미터가 객체 속에 산다** — `Gain(2.0)` 과
  `Gain(5.0)` 은 같은 코드, 다른 물건
- 규약: `output(t, u, x) -> 튜플`. u 는 입력 튜플, 출력도 튜플 (포트 여러 개 대비)
- `Block` 베이스 + 상속: 공통(이름, 포트)은 부모가, 고유(수식)는 자식이
