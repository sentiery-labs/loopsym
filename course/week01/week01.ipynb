목표: **신호 = 시간의 함수** 라는 감각. `step(t)`, `ramp(t)`, `sine(t)` 를 만든다.

> 15주 뒤 당신의 다이어그램이 드론을 날린다 — 오늘은 그 첫 픽셀이다.

## 이론 요약
- 시뮬레이션 = "시간 t 를 조금씩 흘리며 시스템의 값을 계산하는 것"
- 신호는 `f(t) -> 값` 인 함수로 표현할 수 있다
- 파이썬 기초: 변수, 함수 정의(`def`), 기본 인자, 리스트, `for`

````python
# 완성된 미래를 미리 구경 — 정식 blocksim 으로 PI 폐루프 (읽기만!)
import sys; sys.path.insert(0, '../..')
from blocksim import *
bd = Diagram()
ref = bd.add(Step(t_on=1.0)); err = bd.add(Sum('+-'))
ctrl = bd.add(PyFunction('3.0*u0', n_in=1)); plant = bd.add(TransferFunction([1],[1,2,1]))
sc = bd.add(Scope(n_in=2, labels=['ref','y']))
bd.connect(ref, err[0]); bd.connect(plant, err[1]); bd.connect(err, ctrl)
bd.connect(ctrl, plant); bd.connect(ref, sc[0]); bd.connect(plant, sc[1])
run(bd, 10.0, 0.01)
import matplotlib.pyplot as plt
for i, lab in enumerate(sc.labels): plt.plot(sc.t, sc.data[i], label=lab)
plt.legend(); plt.grid(True); plt.title('15주 뒤 당신이 이해하게 될 그림')
````
