## 이론 요약
- **덕타이핑**: 엔진은 `output()` 만 부른다 — 무슨 블록인지 모른다
- 문자열 파라미터가 포트 수를 정한다: `Sum('+-')` → n_in=2 (Simulink 동일)
- `PyFunction`: 식 문자열을 `compile`/`eval` — MATLAB Function 블록의 원리
- 안전한 eval: `{"__builtins__": {}}` 로 위험한 이름 차단

````python
import importlib, myblocks; importlib.reload(myblocks)
s = myblocks.Sum('+-')
print('Sum +- (3,1) →', s.output(0.0, (3.0, 1.0), ()))      # (2.0,)
f = myblocks.PyFunction('6.0*(u0-u1) - 2.0*u2', n_in=3)
print('PD 식 →', f.output(0.0, (0.8, 0.3, 0.1), ()))
````
