**이번 주에 완성되는 엔진 조각**
```
블록(Integrator 추가) + Diagram(완성) → [솔버: 오일러/RK4] → 폐루프
                                          ▲ 여기 — 엔진의 심장
```

## 이론 요약
- `Integrator`: y = x, dx/dt = u — 상태를 가진 첫 블록 (feedthrough=False)
- 오일러: `x += dt·f(t,x)` — 오차 ∝ dt (직접 측정해 본다)
- RK4: 기울기 4개의 가중 평균 — 오차 ∝ dt⁴ (같은 dt 로 10⁶배 정확)
- 진동계에서 오일러는 에너지가 불어나 발산한다 — 안정성은 정확도와 다른 문제

````python
import importlib, myblocks, mydiagram, mysim
for m in (myblocks, mydiagram, mysim): importlib.reload(m)
import math, matplotlib.pyplot as plt
# dx/dt = -x, x(0)=1  →  정답 e^-t
def decay(method, dt):
    bd = mydiagram.Diagram()
    z = bd.add(myblocks.Constant(0.0)); neg = bd.add(myblocks.Sum('+-'))
    it = bd.add(myblocks.Integrator(x0=1.0)); sc = bd.add(myblocks.Scope())
    bd.connect(z, neg[0]); bd.connect(it, neg[1]); bd.connect(neg, it); bd.connect(it, sc)
    mysim.simulate(bd, 1.0, dt, method=method)
    return abs(sc.data[0][-1] - math.exp(-1))
for dt in (0.1, 0.01, 0.001):
    print(f'dt={dt}:  euler 오차 {decay("euler", dt):.2e}   rk4 오차 {decay("rk4", dt):.2e}')
````
